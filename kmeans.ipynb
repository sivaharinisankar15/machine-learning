```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
df = pd.read_csv(r"C:\Users\Harini.S\Desktop\ml\BankChurners.csv")
cols = ["Credit_Limit", "Total_Trans_Amt", "Total_Trans_Ct", "Total_Revolving_Bal", "Avg_Utilization_Ratio"]
X = StandardScaler().fit_transform(df[cols])
print(X.shape)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

for k in range(2, 8):
  km = KMeans(n_clusters=k, n_init=10, random_state=42).fit(X)
  print(k, km.inertia_, silhouette_score(X, km.labels_))

km = KMeans(n_clusters=4, n_init=10, random_state=42).fit(X)
df["segment"] = km.labels_
print(df.groupby("segment")[cols].mean())
```
